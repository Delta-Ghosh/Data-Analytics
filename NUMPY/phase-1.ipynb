## Numpy array and basics
learn about numpy

```python
import numpy as np
```

## creating array 

```python
arr_1 = np.array([1,2,3,4])
print("1d array:", arr_1)

arr_2 = np.array([[1,2,3],[4,5,6]])
print("2d array", arr_2)
```

## list vs array 

```python
py_list = [1,2,3]
print("python list multiplication", py_list*2)

py_arr = np.array ([1,2,3])
print("python array multiplication", py_arr*2)
```

## Creating array from scratch


```python
zeros = np.zeros([3,3])
print("zeros arrays:\n", zeros)

ones = np.ones([2,3])
print("one array:\n", ones)

full = np.full([2,2], 5)
print("full array:\n", full)

random = np.random.random([2,3])
print("random array:\n", random)

sequence = np.arange(0,11,1)
print("sequence array:\n", sequence)
```

## Vector, Matrix, Tensor

```python
vector = np.array([1,2,3])
print("vector: ", vector)

matrix = np.array([[1,2,3],
                   [4,5,6]])
print("Matrix: ", matrix)

tensor = np.array([[[1,2],[3,4]],
                    [[5,6], [7,8]]])
print("tesnor:", tensor)
```

## Array properties 

```python
arr = np.array([[1,2,3],
                [2,3,4]])
print("Shape: ", arr.shape)
print("Dimension: ", arr.ndim)
print("Size: ", arr.size)
print("DataType: ", arr.dtype)
```

## Array Reshaping

```python
arr = np.arange(12)
print("Original array ", arr)

reshaped = arr.reshape((3,4))
print("Reshape array ", reshaped)

flattened = reshaped.flatten()
print("Flattened array ", flattened)

# returns view, instead of copy
raveled = reshaped.ravel()
print("Raveled array ", raveled)

transpose = reshaped.T
print("Transpose array ", transpose)



```
